```python
import pandas as pd
import yfinance as yf
import os

START = "2019-01-01"
END   = "2024-12-31"

os.makedirs("../data/raw", exist_ok=True)
print("✅ Config OK")
```

```python
fx_df = yf.download(["EURUSD=X", "JPY=X"], start=START, end=END)["Close"]
fx_df.columns = ["eurusd", "usdjpy"]
fx_df.index.name = "date"
fx_df.dropna().to_csv("../data/raw/fx_data.csv")
print(fx_df.tail())
```

```python
tickers = ["SPY","XLF","XLE","XLK","XLV","XLI","XLY","XLP","XLU","XLB","XLRE","XLC"]
eq_df = yf.download(tickers, start=START, end=END)["Close"]
eq_df.index.name = "date"
eq_df.dropna().to_csv("../data/raw/equities_data.csv")
print(eq_df.tail())
```

```python
rates_tickers = {"y10": "^TNX", "y30": "^TYX", "vix": "^VIX"}
rates_df = pd.DataFrame({
    name: yf.download(ticker, start=START, end=END)["Close"].squeeze()
    for name, ticker in rates_tickers.items()
})
rates_df.index.name = "date"
rates_df.dropna().to_csv("../data/raw/rates_data.csv")
print(rates_df.tail())
```
