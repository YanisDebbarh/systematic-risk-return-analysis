## 1. Macro Context

The 2019-2024 period covers three distinct regimes:
- **2019-2020:** Zero interest rate policy (ZIRP), COVID shock, emergency Fed cuts
- **2021-2022:** Inflation surge, fastest Fed hiking cycle in 40 years (+525bps)
- **2023-2024:** Normalization, rate plateau, equity rally driven by AI theme

Key observations:
- USD strengthened significantly vs EUR through 2022 (EUR/USD -15%)
- UST 10Y moved from 0.5% (Aug 2020) to 5.0% (Oct 2023)
- VIX spiked to 80 in March 2020, normalized to 12-15 by 2024


## 2. FX Analysis

### EUR/USD
- Trend: Bearish 2021-2022 driven by USD strength (Fed/ECB policy divergence)
- EUR/USD hit lows of ~1.035 in Sept 2022 as Fed hiked aggressively
- Partial recovery in 2023-2024 as ECB caught up with rate hikes
- **Realized Vol:** 7.06% annualized — structurally low vs equities
- **Sharpe Ratio:** -0.89 → euro underperformed risk-free rate over period
- **Max Drawdown:** -22.24%

### USD/JPY
- BoJ maintained ultra-loose policy while Fed hiked → massive carry trade
- USD/JPY rallied from 102 to 152 (2021-2022) — 50% move
- **Realized Vol:** 8.80% annualized
- **Sharpe Ratio:** 0.14 → modest positive return
- **Max Drawdown:** -14.75%

### Key Signal — IV vs RV Gap
- VIX (implied vol proxy) averaged 18-20 during 2023-2024
- EUR/USD realized vol stayed below 7%
- **Gap = ~11-13 vol points → vol structurally overpriced on EUR/USD**
- Trading implication: selling EUR/USD straddles or strangles was systematically profitable


## 3. Rates Analysis

### Yield Curve Dynamics
- 2020: Curve bull steepened as Fed cut to 0% (2Y anchored, 10Y fell less)
- 2022: Aggressive bear flattening then inversion (2s10s deeply negative)
- 2023-2024: Gradual normalization, curve still partially inverted

### Key Signal — Bear Steepening Detected
- 10Y rose faster than 2Y in late 2023 → classic bear steepening
- **Implication:** Long end selling pressure, inflation premium repricing
- **Trade idea:** Pay 10Y, receive 2Y (steepener) — positive carry in inversion

### DV01 Sensitivity (from Excel model)
- 10Y Treasury ($1M notional): DV01 = $850
- 30Y Treasury ($1M notional): DV01 = $1,800
- Hausse 50bps: P&L = -$42,500 (10Y), -$90,000 (30Y)
- **Key takeaway:** Long duration = significant rate risk in hiking cycle


## 4. Equities & Cross-Asset

### Sector Leadership
- **2020:** XLK (Tech) outperformed, XLE (Energy) collapsed (-40%)
- **2022:** XLE only positive sector (+60%), XLK worst (-35%)
- **2023-2024:** XLK dominated again driven by AI theme (NVIDIA effect)

### Cross-Asset Correlations
- SPY vs Sectors: 0.82-0.93 → highly correlated, limited diversification
- VIX vs SPY: -0.70 → fear index reliably inverse to equity performance
- FX vs Equities: ~0.03 → FX decorrelated from equities over period

### Key Signal — Sector Rotation
- Energy → Tech rotation from 2022 to 2024
- Defensive sectors (XLU, XLP) underperformed in risk-on 2023-2024
- **Implication:** Momentum in XLK, mean-reversion opportunity in XLE


## 5. Trading Conclusions & Trade Ideas

### Trade 1 — Short EUR/USD Vol
- **Signal:** IV-RV gap of 11-13 vol points on EUR/USD
- **Structure:** Sell 1M EUR/USD strangle (25-delta)
- **Rationale:** Realized vol structurally below implied → theta positive
- **Risk:** Macro shock (Fed surprise, geopolitical event)

- **Structure:** Pay 10Y swap, receive 2Y swap
- **Rationale:** Long end repricing inflation premium, Fed cuts anchoring 2Y
- **Risk:** Recession → bull flattening reverses trade

### Trade 3 — Sector Momentum

- **Structure:** Long XLK, short XLE (pairs trade)
